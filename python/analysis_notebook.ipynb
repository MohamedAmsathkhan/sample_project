```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import sqlite3

# Load data from SQLite database into Pandas DataFrame
conn = sqlite3.connect('../data/ecommerce.db')
df = pd.read_sql_query('SELECT * FROM sales', conn)
conn.close()

df.head()
```

```python
# Set theme for Seaborn
sns.set_theme(style='darkgrid')

# Plot Revenue by Category Bar Chart
plt.figure(figsize=(8, 4))
sns.barplot(data=df, x='category', y='total_amount', palette='viridis', hue='category')
plt.title('Total Sales Revenue by Category ($)', fontsize=14)
plt.xlabel('Category')
plt.ylabel('Revenue ($)')
plt.show()
```

```python
# Plot Order Status Distribution
plt.figure(figsize=(5, 5))
df['order_status'].value_counts().plot.pie(autopct='%1.1f%%', colors=['#4CAF50', '#FF5722'], startangle=90)
plt.title('Order Status (Completed vs Returned)')
plt.ylabel('')
plt.show()
```
